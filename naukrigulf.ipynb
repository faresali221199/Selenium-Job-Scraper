```python
import time
import csv
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

options = webdriver.ChromeOptions()
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")

driver = webdriver.Chrome(options=options)
driver.maximize_window()
wait = WebDriverWait(driver, 15)

all_jobs = []
num_pages = 3

for page_number in range(1, num_pages + 1):
    print("=" * 60)
    print(f"جاري جلب البيانات من الصفحة رقم {page_number}")
    
    url = f"https://www.naukrigulf.com/data-engineer-jobs-{page_number}"
    
    try:
        driver.get(url)
        time.sleep(3)
        
        job_cards = wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "div.srp-tuple")))
        print(f"عدد الوظائف في هذه الصفحة: {len(job_cards)}")

        for card in job_cards:
            try:
                title_elem = card.find_element(By.CSS_SELECTOR, "a.info-position")
                job_title = title_elem.text.strip()
                job_url = title_elem.get_attribute("href")

                try:
                    location = card.find_element(By.CSS_SELECTOR, "li.info-loc").text.strip()
                except:
                    location = "Not found"

                try:
                    experience = card.find_element(By.CSS_SELECTOR, "li.info-exp").text.strip()
                except:
                    experience = "Not found"

                try:
                    company_name = card.find_element(By.CSS_SELECTOR, ".info-org").text.strip()
                except:
                    company_name = "Not found"

                all_jobs.append({
                    "title": job_title,
                    "company": company_name,
                    "location": location,
                    "experience": experience,
                    "url": job_url
                })
            except Exception as e:
                continue

    except Exception as e:
        continue

print("\n" + "=" * 60)
print(f"تم العثور على إجمالي: {len(all_jobs)} وظيفة")
print("=" * 60)


print("\nجاري الدخول إلى الوظائف وجلب الوصف الكامل...\n")

for index, job in enumerate(all_jobs, start=1):
    print(f"جاري معالجة وظيفة ({index}/{len(all_jobs)}): {job['title']}")
    
    try:
        driver.get(job["url"])
        time.sleep(2) 
        
        full_desc_elem = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "article.job-description")))
        full_description = full_desc_elem.text.strip()

    except Exception as e:
        full_description = "لم يتم العثور على الوصف الكامل"

    job['full_description'] = full_description

driver.quit()


print("\nجاري حفظ جميع البيانات في ملف CSV...")

with open('data_engineer_jobs.csv', mode='w', newline='', encoding='utf-8-sig') as file:
    writer = csv.writer(file)
    
  
    writer.writerow(['Job Title', 'Company Name', 'Job Location', 'Required Experience', 'Full Job Description'])
    

    for job in all_jobs:
        writer.writerow([
            job['title'], 
            job['company'], 
            job['location'], 
            job['experience'], 
            job['full_description']
        ])

print("\n تمت العملية بنجاح!  ")
```
